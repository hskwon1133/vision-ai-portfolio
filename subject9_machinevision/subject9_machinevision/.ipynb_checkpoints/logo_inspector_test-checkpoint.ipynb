````python
#실습 1: 사진을 읽고 로고 ROI 잘라 보기
from pathlib import Path
import cv2
import matplotlib.pyplot as plt

img_path = Path(
    "samples/day04/source/full_front.jpg"
)
image = cv2.imread(str(img_path))
print(image is None)      # False가 나와야 읽기 성공
print(image.shape)        # (높이, 너비, 채널)
````

````python
x1, y1, x2, y2 = 510, 870, 950, 1080 ## y2 = 1080으로 지정했으나 img 크기는 width 1068까지임
reference = image[y1:y2, x1:x2]

print("내가 요청한 크기:", (y2 - y1, x2 - x1))
print("실제로 잘린 크기:", reference.shape[:2])
````

1. 왜 내가 요청한 크기와 실제 잘린 크기가 다를까요?
- y2 = 1080으로 요청되었으나, 이미지 width는 1068임.
- -12만큼의 gap을 이미지 width에서 빼줌.
- **중요** ROI 좌표를 정한 뒤 reference.shape를 꼭 print해서 확인필수

````python
view = image.copy()
cv2.rectangle(view, (x1, y1), (x2, y2), (0, 255, 0), 4)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1); plt.imshow(cv2.cvtColor(view, cv2.COLOR_BGR2RGB)); plt.title("full_front + ROI")
plt.subplot(1, 2, 2); plt.imshow(cv2.cvtColor(reference, cv2.COLOR_BGR2RGB)); plt.title("reference")
plt.show()
````

````python
# 실습 2: shift 이미지를 만들어 로고가 잘리는 걸 확인하기

import numpy as np

h, w = image.shape[:2]

#픽셀 하나하나를 어디로 옮길지 정하는 공식표
M = np.float32([[1, 0, 35],    # x 방향: 오른쪽으로 35
                [0, 1, 20]])   # y 방향: 아래로 20

#warpAffine은 2x3행렬구조를 가짐
#BORDER_REPLICATE: OpenCV등 이미지 처리 라이브러리에서 이미지의 가장자리(최외곽)픽셀값을 복제하여 바깥쪽 패딩(border)를 채우는 방식
shift = cv2.warpAffine(image, M, (w, h), borderMode=cv2.BORDER_REPLICATE) 
print(shift.shape)   # 사진 크기는 그대로일까?
````

### 📌 warpAffine 정리: 액자 비유

**이동 행렬 M**
```
M = [[1, 0, tx],    새 x = x + tx  (+면 오른쪽)
     [0, 1, ty]]    새 y = y + ty  (+면 아래, OpenCV는 y가 아래로 증가)
```
- 앞의 2×2 `[[1,0],[0,1]]` = 모양 그대로 (크기·각도 변화 없음)
- `np.float32`로 만들어야 함 (warpAffine이 실수형 행렬을 요구)

**출력 크기 (w, h) = 고정된 액자**
- 사진을 밀어도 액자 크기는 그대로
- 밀려난 쪽 → 액자 밖으로 나가서 **사라짐**
- 반대쪽 → **빈 띠** 생김 → `borderMode`로 채움

| borderMode | 빈 곳 채우는 방법 |
|---|---|
| `BORDER_CONSTANT` | 0(검정)으로 채움 → 검은 띠 |
| `BORDER_REPLICATE` | 끝 픽셀 복사 → 자연스러움 |

**REPLICATE를 쓰는 이유:** 검은 띠 경계의 날카로운 직선이 ORB/SIFT에서 가짜 특징점을 만들기 때문

**실습 결과:** 로고가 사진 맨 아래에 붙어 있어서(y=870~1067), 아래로 20 밀면 로고 아래 20줄이 액자 밖으로 잘림 → Template Matching shift 실패 원인

### 📌 요청 크기 ≠ 실제 크기
- 사진 높이 1068인데 ROI를 y=870~1080으로 요청 → 실제로는 198줄만 잘림 (210 아님)
- NumPy 슬라이싱 `a[시작:끝]`은 범위를 넘어도 **에러 없이** 있는 데까지만 자름
- 교훈: ROI를 자른 뒤 `print(reference.shape)`로 꼭 확인하기

````python
#실습 3: 회전 각도 바꿔 보기 (5°, 15°, 30°)
#셀 1: 회전 행렬 만들기와 들여다보기
h, w = image.shape[:2]
center = (w / 2, h / 2)

R = cv2.getRotationMatrix2D(center, 15, 1.0)   # (중심, 각도, 배율)
print(np.round(R, 3))
````

````python
#셀 2: 각도 3개를 고정 액자로 돌려 보기

angles = [0, 5, 15, 30]

plt.figure(figsize=(20, 4))
for i, ang in enumerate(angles):
    R = cv2.getRotationMatrix2D(center, ang, 1.0)
    rot = cv2.warpAffine(image, R, (w, h), borderMode=cv2.BORDER_REPLICATE)
    plt.subplot(1, 4, i + 1)
    plt.imshow(cv2.cvtColor(rot, cv2.COLOR_BGR2RGB)); plt.title(f"rot {ang} (fixed frame)")
plt.show()
````

````python
#셀 3: 액자를 키워서 잘리지 않게 돌리기 (강사님 24_day04_prepare.py 방식)
def rotate_keep_all(img, angle):
    h, w = img.shape[:2]
    R = cv2.getRotationMatrix2D((w / 2, h / 2), angle, 1.0)
    cos, sin = abs(R[0, 0]), abs(R[0, 1])
    new_w = int(w * cos + h * sin)       # 돌린 사진을 다 담는 새 액자 크기
    new_h = int(w * sin + h * cos)
    R[0, 2] += new_w / 2 - w / 2         # 새 액자 가운데로 옮기기
    R[1, 2] += new_h / 2 - h / 2
    return cv2.warpAffine(img, R, (new_w, new_h), borderMode=cv2.BORDER_REPLICATE)

plt.figure(figsize=(20, 4))
for i, ang in enumerate(angles):
    rot = rotate_keep_all(image, ang)
    print(ang, "도 → 새 크기:", rot.shape[:2])
    plt.subplot(1, 4, i + 1)
    plt.imshow(cv2.cvtColor(rot, cv2.COLOR_BGR2RGB)); plt.title(f"rot {ang} (expanded)")
plt.show()

````

````python
#실습 4: 크기 변화와 Template Matching
#셀 1: 크기 행렬 들여다보기
h, w = image.shape[:2]
S = cv2.getRotationMatrix2D((w / 2, h / 2), 0, 0.90)   # 각도 0, 배율 0.9
print(np.round(S, 3))
````

````python
#셀 2: 축소된 사진에서 로고가 어디로, 얼마나 작아졌는지 계산하기
corners = np.array([[510, 870, 1],      # 로고 왼쪽 위 (x, y, 1)
                    [950, 1068, 1]]).T  # 로고 오른쪽 아래
new = (S @ corners).T                   # 행렬 곱 = 실습 2에서 손으로 했던 계산
print(np.round(new, 1))

nw = new[1, 0] - new[0, 0]
nh = new[1, 1] - new[0, 1]
print("원래 로고 크기:", 440, "x", 198)
print("축소 후 로고 크기:", round(nw), "x", round(nh))
````

````python
#셀 3: 원본 크기 템플릿으로 찾기 (교재 방식)
#각 위치에서 "얼마나 비슷한가?" 계산
s_img = cv2.imread("samples/day04/targets/target_scale90.jpg")
s_gray = cv2.cvtColor(s_img, cv2.COLOR_BGR2GRAY)
ref_gray = cv2.cvtColor(image[870:1068, 510:950], cv2.COLOR_BGR2GRAY)

score_map = cv2.matchTemplate(s_gray, ref_gray, cv2.TM_CCOEFF_NORMED)
_, max_score, _, max_loc = cv2.minMaxLoc(score_map)
print("score:", round(max_score, 4), "loc:", max_loc)
````
